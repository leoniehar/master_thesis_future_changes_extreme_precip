# bias plots (QQ plot kinda)

```python
sns.set_context("talk")

icon = icon_images
obs = obs_mean_images

n_times = len(icon.time)

ncols = 4
nrows = int(np.ceil(n_times / ncols))

fig, axes = plt.subplots(nrows, ncols, figsize=(4*ncols, 4*nrows))
axes = np.array(axes).flatten()

for i, t in enumerate(icon.time):

    x = obs.sel(time=t).values.flatten()
    y = icon.sel(time=t).values.flatten()

    mask = ~np.isnan(x) & ~np.isnan(y)
    x = x[mask]
    y = y[mask]

    # sort → QQ logic
    x_sorted = np.sort(x)
    y_sorted = np.sort(y)

    # match lengths (important for QQ stability)
    n = min(len(x_sorted), len(y_sorted))
    x_sorted = x_sorted[:n]
    y_sorted = y_sorted[:n]

    axes[i].plot(x_sorted, y_sorted, ".", markersize=1)

    # 1:1 reference line
    lims = [0, max(np.nanmax(x_sorted), np.nanmax(y_sorted))]
    axes[i].plot(lims, lims, "r")

    axes[i].set_xlim(lims) # lims 
    axes[i].set_ylim(lims) # lims

    axes[i].set_title(str(t.values)[:16])
    axes[i].set_xlabel("OBS quantiles")
    axes[i].set_ylabel("ICON quantiles")

# hide unused axes
for j in range(i+1, len(axes)):
    axes[j].axis("off")

fig.suptitle("QQ-kinda comparison: ICON vs OBS mean", fontweight="bold")
plt.tight_layout()
plt.show()
```
