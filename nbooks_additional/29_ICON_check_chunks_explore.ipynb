```python
import zarr
import numpy as np
import pandas as pd

store = zarr.open("https://s3.eu-dkrz-1.dkrz.cloud/nextgems/rechunked_ngc4008/ngc4008_PT3H_9.zarr", mode="r")

# Check chunk shape of tas — first dimension is time
print(store["tas"].chunks)   # e.g. (10, 128, 256) → chunk size along time = 10
print(store["tas"].shape)    # total shape

# Decode the time axis
time = store["time"][:]
time_decoded = pd.to_datetime(time, unit="s", origin="unix")  # adjust unit/origin to your encoding
print(time_decoded[:5])
```

```python
chunk_t = store["tas"].chunks[0]  # time chunk size
for chunk_idx in [40, 41]:
    i_start = chunk_idx * chunk_t
    i_end   = (chunk_idx + 1) * chunk_t
    print(f"chunk {chunk_idx}: {time_decoded[i_start]} → {time_decoded[i_end-1]}")
```
