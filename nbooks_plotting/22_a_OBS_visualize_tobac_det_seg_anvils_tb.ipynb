# Basic summary of concept plot

```python
import matplotlib.patches as mpatches
import numpy as np
from matplotlib.transforms import Affine2D


date_title = '2023-06-06 18:00 UTC'
sns.set_context("talk" ,font_scale =2)

TROPICS = [-180, 180, -30, 30]  # lon_min, lon_max, lat_min, lat_max
BOX_COLOR = "red"


def _fig_size(extent, ncols, nrows, width=12, extra=0.5):
    """Figure size from the lon/lat extent of one panel."""
    lon_min, lon_max, lat_min, lat_max = extent
    aspect = (lat_max - lat_min) / (lon_max - lon_min)
    panel_w = width / ncols
    height = nrows * panel_w * aspect + extra
    return width, max(4, min(height, 20))


def mark_region(ax, extent, color=BOX_COLOR):
    """Draw a box around the zoom region."""
    lon_min, lon_max, lat_min, lat_max = extent
    ax.add_patch(mpatches.Rectangle(
        (lon_min, lat_min), lon_max - lon_min, lat_max - lat_min,
        fill=False, edgecolor=color, linewidth=2.5,
        transform=ccrs.PlateCarree(), zorder=10))


def plot_tropics(tb, precip, date_title, zoom_extent=None, extent=TROPICS):
    """Global BT (top) and precipitation (bottom), optional zoom box."""
    fig, axes = plt.subplots(
        2, 1,
        figsize=_fig_size(extent, ncols=1, nrows=2),
        subplot_kw={"projection": ccrs.PlateCarree()},
        constrained_layout=True,
    )
    plots_mapping.domain_plot_tb(
        tb, domain=extent, left_grid_label=True, bottom_grid_label=False,
        title="a) Brightness Temperature c) d)", fig=fig, ax=axes[0])
    plots_mapping.domain_plot_precip(
        precip, domain=extent, left_grid_label=True, bottom_grid_label=True,
        title="b) Precipitation", fig=fig, ax=axes[1])

    fig.get_layout_engine().set(h_pad=0.02, hspace=0.0, w_pad=0.02)

    if zoom_extent is not None:
        for ax in axes:
            mark_region(ax, zoom_extent)

    fig.suptitle(date_title, fontweight="bold", x=0.5)
    # move colorbars closer to the maps
    for cax in fig.axes:
        if cax not in axes:
            cax._colorbar_info["pad"] = 0.00  # default is ~0.05
    return fig, axes


def plot_zoom(tb, precip, date_title, extent):
    """Zoom on one region: BT (left) and precipitation (right)."""
    fig, axes = plt.subplots(
        1, 2,
        figsize=_fig_size(extent, ncols=2, nrows=1),
        subplot_kw={"projection": ccrs.PlateCarree()},
        constrained_layout=True,
    )
    plots_mapping.domain_plot_tb(
        tb, domain=extent, left_grid_label=True, bottom_grid_label=True,
        title="a) Brightness Temperature", fig=fig, ax=axes[0])
    plots_mapping.domain_plot_precip(
        precip, domain=extent, left_grid_label=True, bottom_grid_label=True,
        title="b) Precipitation", fig=fig, ax=axes[1])

    fig.suptitle(date_title, fontweight="bold", x=0.5)

           
    return fig, axes


zoom_extent = [76.5, 82, 11, 14]

fig, axes = plot_tropics(tb, precip, date_title, zoom_extent)
fig.savefig(os.path.join(plot_dir,f"intro_fig_global_map_{date}.pdf"),dpi=150, bbox_inches='tight')
plt.show()

fig_zoom, axes_zoom = plot_zoom(tb, precip, date_title, zoom_extent)
fig_zoom.savefig(os.path.join(plot_dir,f"intro_fig_zoom_{date}.png"),dpi=150, bbox_inches='tight')

plt.show()
```

```python
def render_panel(plot_func, data, extent, title="", labels=True,
                 figsize=(5, 3.4), dpi=450):
    """Render one map panel to an RGBA array, without colorbar."""
    fig, ax = plt.subplots(
        figsize=figsize, dpi=dpi,
        subplot_kw={"projection": ccrs.PlateCarree()},
        constrained_layout=True)
    fig.patch.set_alpha(0)
    plot_func(data, domain=extent, left_grid_label=labels,
              bottom_grid_label=labels, title=title, fig=fig, ax=ax)

    fig.canvas.draw()                    # let the layout settle first
    fig.set_layout_engine("none")        # then freeze it
    for a in fig.axes:
        if a is not ax:
            a.set_visible(False)         # hide the colorbar
    fig.canvas.draw()

    img = np.asarray(fig.canvas.buffer_rgba()).copy()
    plt.close(fig)
    return img

def plot_tilted(imgs, angle=30, squash=0.5, gap=1):
    """Draw images as skewed planes, stacked top to bottom."""
    h, w = imgs[0].shape[:2]
    step = h * squash * gap
    fig, ax = plt.subplots(figsize=(15, 15), constrained_layout=True)
    for i, img in enumerate(imgs):
        tr = (Affine2D().scale(1, squash).skew_deg(angle, 0)
              .translate(0, -i * step) + ax.transData)
        ax.imshow(img, extent=[0, w, 0, h], transform=tr)
    ax.set_xlim(0, w + h * squash * np.tan(np.radians(angle)))
    ax.set_ylim(-(len(imgs) - 1) * step, h * squash)
    ax.set_aspect("equal")
    ax.axis("off")
    return fig, ax


img_tb = render_panel(plots_mapping.domain_plot_tb, tb, zoom_extent)
img_pr = render_panel(plots_mapping.domain_plot_precip, precip, zoom_extent)

fig_tilt, ax_tilt = plot_tilted([img_tb, img_pr], angle=45)
fig_tilt.savefig(os.path.join(plot_dir, f"intro_fig_zoom_tilted_{date}.png"),
                 dpi=300, transparent=True, bbox_inches="tight")
plt.show()
```
