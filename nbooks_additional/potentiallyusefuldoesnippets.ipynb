## find equivalent grid size (from healpix

```python
# von Fabian
d_lon_lat = precip.isel(time = [itime,itime+1], cell=idx)

dlon = d_lon_lat.lon.diff('lon').mean().data
dlat = d_lon_lat.lat.diff('lat').mean().data

lat0 = d_lon_lat.lat.mean('lat').mean().data

R_earth = 6371e3

dlambda = np.deg2rad( dlon )
dphi = np.deg2rad( dlat )
phi0 = np.deg2rad( lat0 )

dA = R_earth**2 * np.cos( phi0 ) * dlambda * dphi

dX = np.sqrt( dA )
print(dX)
```

```python
def params_figure_from_json(json_path):
    """
    Loads a params JSON and saves a matching figure next to it.
    """
    json_path = Path(json_path)
    with open(json_path, 'r') as f:
        params = json.load(f)

    blocks = []
    for name, value in params.items():
        if value is None:
            continue
        if isinstance(value, dict):
            formatted = plots_mapping.format_dict(value)
        else:
            formatted = str(value)
        blocks.append(f"{name}:\n{formatted}")

    param_text = "\n\n".join(blocks)
    n_lines = param_text.count("\n") + 1
    fig_height = max(3, n_lines * 0.22)

    fig, ax = plt.subplots(figsize=(3, fig_height))
    ax.axis("off")
    ax.text(
        0.03, 0.97,
        param_text,
        transform=ax.transAxes,
        fontsize=12,
        verticalalignment="top",
        horizontalalignment="left",
        family="monospace",
        bbox=dict(
            boxstyle="round,pad=0.6",
            facecolor="white",
            alpha=1.0,
            edgecolor="gray",
            linewidth=0.8,
        ),
    )

    tag = json_path.stem  # e.g. "params_anvil_v1"
    fig.suptitle(tag,fontsize=12, fontweight="bold", y=1.01)
    fig.tight_layout()

    
    plt.show(fig)
    return 
    

# call
json_path = savedir / f"params_anvil_{param_tag}.json"
params_figure_from_json(json_path =json_path)
# save
out_path = json_path.with_suffix(".png")
fig.savefig(out_path, dpi=150, bbox_inches="tight")
```
