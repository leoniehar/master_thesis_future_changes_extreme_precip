```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python

sns.set_context("talk")

x = np.linspace(0, 2 * np.pi, 20)

color_precip_icon = "dodgerblue"
color_precip_obs  = "darkblue"
color_tb_icon     = "orangered"
color_tb_obs      = "firebrick"
color_iwp_icon = "violet"
color_iwp_obs  = "mediumorchid"

fig, ax = plt.subplots(figsize=(8,5))

ax.plot(x, np.sin(x) + 3.0, color=color_precip_icon, marker='o', label="Precip-ICON")
ax.plot(x, np.sin(x) + 2.5, color=color_precip_obs,  marker='o', label="Precip-OBS")
ax.plot(x, np.sin(x) + 2.0, color=color_tb_icon,     marker='o', label="Tb-ICON")
ax.plot(x, np.sin(x) + 1.5, color=color_tb_obs,      marker='o', label="Tb-OBS")
ax.plot(x, np.sin(x) + 1.0, color=color_iwp_icon,    marker='o', label="IWP-ICON")
ax.plot(x, np.sin(x) + 0.5, color=color_iwp_obs,     marker='o', label="IWP-OBS")

ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_title("Color palette preview", fontweight="bold")
ax.legend(frameon=True)
sns.despine(ax=ax)
plt.tight_layout()
plt.show()

fig, ax = plt.subplots(figsize=(8,5))

#ax.plot(x, np.sin(x) + 3.0, color=color_precip_icon, marker='o', label="Precip-ICON")
ax.plot(x, np.sin(x) + 2.5, color=color_precip_obs,  marker='o', label="Precip-OBS")
#ax.plot(x, np.sin(x) + 2.0, color=color_tb_icon,     marker='o', label="Tb-ICON")
ax.plot(x, np.sin(x) + 1.5, color=color_tb_obs,      marker='o', label="Tb-OBS")
#ax.plot(x, np.sin(x) + 1.0, color=color_iwp_icon,    marker='o', label="IWP-ICON")
ax.plot(x, np.sin(x) + 0.5, color=color_iwp_obs,     marker='o', label="IWP-OBS")

ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_title("Color palette preview", fontweight="bold")
ax.legend(frameon=True)
sns.despine(ax=ax)
plt.tight_layout()
plt.show()

fig, ax = plt.subplots(figsize=(8,5))

ax.plot(x, np.sin(x) + 3.0, color=color_precip_icon, marker='o', label="Precip-ICON")
#ax.plot(x, np.sin(x) + 2.5, color=color_precip_obs,  marker='o', label="Precip-OBS")
ax.plot(x, np.sin(x) + 2.0, color=color_tb_icon,     marker='o', label="Tb-ICON")
#ax.plot(x, np.sin(x) + 1.5, color=color_tb_obs,      marker='o', label="Tb-OBS")
ax.plot(x, np.sin(x) + 1.0, color=color_iwp_icon,    marker='o', label="IWP-ICON")
#ax.plot(x, np.sin(x) + 0.5, color=color_iwp_obs,     marker='o', label="IWP-OBS")

ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_title("Color palette preview", fontweight="bold")
ax.legend(frameon=True)
sns.despine(ax=ax)
plt.tight_layout()
plt.show()
```
