# quick video

```python
import pandas as pd
import imageio.v2 as imageio
import os
import glob
```

```python
# output stuff
outputdir = "../output/animations/quickplots_segmentation/"
gif_file = os.path.join(outputdir, "quickplots_segmentation_ICON.gif")

# input stuff
inputdir = "../output/plots/quickplots_segmentation_ICON/"
start = pd.to_datetime("2020-06-06-00:00", format="%Y-%m-%d-%H:%M")
end   = pd.to_datetime("2020-06-06-21:00", format="%Y-%m-%d-%H:%M")

# date list
datelist = pd.date_range(start, end, freq="3h").strftime("%Y-%m-%d-%H%M").tolist()

os.makedirs(outputdir, exist_ok=True)
```

```python
# loop for animation
# note: if using imageio v3, duration is in milliseconds — use 500 instead of 0.5
with imageio.get_writer(gif_file, mode="I", duration=1000, loop = 0) as writer:
    for ts in datelist:
        pattern = os.path.join(inputdir, f"*{ts}.png")
        matches = glob.glob(pattern)
        if matches:
            image = imageio.imread(matches[0])  # take first match if multiple
            writer.append_data(image)
        else:
            print(f"File not found: {pattern}")  # was incorrectly referencing undefined `file`

print(f"GIF animation saved to {gif_file}")
```
