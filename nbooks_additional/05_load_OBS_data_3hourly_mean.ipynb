# COMAPRE RUNTIMES
# Snapshot version

```python
repeattimes = 10
overview_snapshot_times = []

for i in range(repeattimes):
    start = time.perf_counter()
    
    # load data (load, allign coords, selects Tb and precip variable)
    tb = handle_OBS_data.load_clean_crop_tb(date,32) # tb and iwp need resampling to coarser grid=> they are loaded with a 2° buffer so resampling at the edges has enough data
    
    # load data
    precip = handle_OBS_data.load_clean_crop_imerg(date,30) # precip is ref grid => AOI is +-30° lat 
    
    # load data
    iwp = handle_OBS_data.load_clean_crop_ccic(date,32)
    
    end = time.perf_counter()
    
    #print(f"\nTotal time: {end - start:.2f} seconds")
    overview_snapshot_times.append(end - start)
```

```python
repeattimes = 10
overview_6images_times = []

for i in range(repeattimes):
    start = time.perf_counter()
    
    tb = handle_OBS_data.load_clean_crop_tb_forpast3hours(date,32)
    
    precip = handle_OBS_data.load_clean_crop_imerg_forpast3hours(date,30)
    
    iwp = handle_OBS_data.load_clean_crop_ccic_forpast3hours(date,32)
    
    end = time.perf_counter()
    #print(f"\nTotal time: {end - start:.2f} seconds")
    overview_6images_times.append(end - start)
```

```python
repeattimes = 10
overview_mean_times = []

for i in range(repeattimes):
    start = time.perf_counter()
    
    tb_mean = handle_OBS_data.get_3_hour_mean_pixelwise(tb, min_images= 5,ref_time= date).compute()
    
    precip_mean= handle_OBS_data.get_3_hour_mean_pixelwise(precip, min_images= 5,ref_time= date).compute()
    
    iwp_mean = handle_OBS_data.get_3_hour_mean_pixelwise(iwp, min_images= 5,ref_time= date).compute()
    
    end = time.perf_counter()
    #print(f"\nTotal time: {end - start:.2f} seconds")
    overview_mean_times.append(end - start)
```
