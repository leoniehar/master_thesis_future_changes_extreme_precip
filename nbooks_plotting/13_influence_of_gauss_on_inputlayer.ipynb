# access influences of gaussian smoothing applied to the OBS preprocessing
- input: Nasa-Tb, IWP and imerg reference grid
- preprocess with and without Gaussian smoothing
- output:
    - map of comparison: with Gaussian smoothing and without Gaussian smoothing
    - histogram (cdf) comparison

```python
import numpy as np
import xarray as xr


import warnings
warnings.filterwarnings("ignore", category=FutureWarning)


import sys
sys.path.append("../tools/")
import handle_ICON_data as ic
import plots_mapping as pm
import handle_OBS_data
import inputlayer


import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import seaborn as sns

import os
```

```python
date = "2023-06-06-1800"


import os
output_directory = "../output/plots"
if not os.path.exists(os.path.join(output_directory)):
    os.makedirs(os.path.join(output_directory,), exist_ok=True)
```

```python
# mean
tb = handle_OBS_data.load_calc_clean_mean_tb(date, buffer = 32, min_images= 5)
precip = handle_OBS_data.load_calc_clean_mean_imerg(date, buffer = 30, min_images= 5)
iwp = handle_OBS_data.load_calc_clean_mean_ccic(date, buffer = 32, min_images= 5)

# modify short names so that match
precip.attrs["short_name"] = "Rain rate"
tb.attrs["short_name"] = "BT"

iwp.attrs["units"] = r"kg m$^{-2}$"
```

```python
def summarized_domain_plot(tb, tb_gaus, iwp, iwp_gaus, title, domain=[-180,180,-30,30],
                           tb_vmin=200, tb_vmed=240, tb_vmax=300,
                           precip_vmin=0, precip_vmax=3,
                           iwp_vmin=1e-3, iwp_vmax=1e1):

    sns.set_context("talk")

    fig, axes = plt.subplots(
        4, 1,
        figsize=(12, 10),
        subplot_kw={"projection": ccrs.PlateCarree()},
        constrained_layout=True
    )

    pm.domain_plot_tb(tb, domain=domain, left_grid_label=True, bottom_grid_label=False,
                   title="",
                   vmin=tb_vmin, vmed=tb_vmed, vmax=tb_vmax,
                   fig=fig, ax=axes[0],shrink = 0.7)

    pm.domain_plot_tb(tb_gaus, domain=domain, left_grid_label=True, bottom_grid_label=False,
                   title="",
                   vmin=tb_vmin, vmed=tb_vmed, vmax=tb_vmax,
                   fig=fig, ax=axes[1],shrink = 0.7)

    pm.domain_plot_iwp(iwp, vmin=iwp_vmin, vmax=iwp_vmax,
                    domain=domain, left_grid_label=True, bottom_grid_label=True,
                    title="",
                    fig=fig, ax=axes[2], shrink = 0.7)

    pm.domain_plot_iwp(iwp_gaus, vmin=iwp_vmin, vmax=iwp_vmax,
                    domain=domain, left_grid_label=True, bottom_grid_label=True,
                    title="", 
                    fig=fig, ax=axes[3], shrink = 0.7)

    axes[0].set_title("a) BT - NN", loc="left", fontweight = "bold")
    axes[1].set_title("b) BT - NN + Gaussian filter", loc="left",fontweight = "bold")
    axes[2].set_title("c) IWP - NN", loc="left",fontweight = "bold")
    axes[3].set_title("d) IWP - NN + Gaussian filter", loc="left",fontweight = "bold")

    fig.suptitle(title, fontweight="bold")

    return fig
```

```python
tb_res = handle_OBS_data.resample_grids_old(tb,precip)  
iwp_res = handle_OBS_data.resample_grids_old(iwp,precip)  

# gauss regrid
tb_gaus_res =handle_OBS_data.resample_grids(tb,precip)  #handle_OBS_data.
iwp_gaus_res = handle_OBS_data.resample_grids(iwp,precip) #handle_OBS_data.


iwp_gaus_res.attrs["units"] = r"kg m$^{-2}$"
```

```python
date_label = pm.title_time_format(iwp_res)

fig = summarized_domain_plot(tb =tb_res, tb_gaus = tb_gaus_res,
                             iwp = iwp_res,iwp_gaus = iwp_gaus_res,
                             title = f"OBS - {date_label} UTC")
```

```python
fig.savefig(os.path.join(output_directory,f"OBS_comparison_withandwithout_gaussian_{date}.pdf"))
```

```python
tb_res = tb_res.expand_dims("time")
tb_gaus_res = tb_gaus_res.expand_dims("time")
tb = tb.expand_dims("time")
```

```python
iwp = iwp.expand_dims("time")
iwp_res = iwp_res.expand_dims("time")
iwp_gaus_res = iwp_gaus_res.expand_dims("time") 
```

# compare raw, nearest neighbour and gauss+nearest

```python
small= 800

tb_bins_small =  np.linspace(190,330,small)

iwp_bins_small = inputlayer.log_bins(low=-2, high=2, n = small)
```

```python
area_weights_raw = inputlayer.calc_area_weights(tb)  
area_weights_res = inputlayer.calc_area_weights(tb_res)  # same grid
```

```python
# tbs
hist_tb = inputlayer.compute_pdf([tb,tb_res, tb_gaus_res], tb_bins_small, labels = ["tb","tb_res","tb_gaus_res"], filter_positive=False)
# add cdf
hist_tb = inputlayer.get_cdf(hist_tb)

# iwp
hist_iwp = inputlayer.compute_pdf([iwp_res,iwp_gaus_res], iwp_bins_small, labels = ["iwp_res","iwp_gaus_res"], filter_positive=False)
# add contribution
#hist_iwp = add_contribution(hist_iwp)
hist_iwp_amount =inputlayer.compute_amount_distribution([iwp_res,iwp_gaus_res], iwp_bins_small, labels = ["iwp_res","iwp_gaus_res"], filter_positive=False,area_weights = area_weights_res)
hist_iwp_amount_raw =inputlayer.compute_amount_distribution([iwp], iwp_bins_small, labels = ["iwp"], filter_positive=False,area_weights = area_weights_raw)

hist_iwp = xr.merge([hist_iwp, hist_iwp_amount,hist_iwp_amount_raw])
# add cdf
hist_iwp = inputlayer.get_cdf(hist_iwp)

```

```python
hist_iwp
```

```python
color_raw="darkgrey"
color_res = "lightblue"
color_gaus_res = "firebrick"
color_iwp_obs  = "mediumorchid"

sns.set_context("talk")
fig, axes = plt.subplots(1, 3, figsize=(12,4))


inputlayer.plot_pdf_pair( ax=axes[0], data_list=[hist_tb["cdf_counts_tb"],hist_tb["cdf_counts_tb_res"], hist_tb["cdf_counts_tb_gaus_res"]], 
               alltimestamps = False, main = True, uncertainty = False,
               percentiles =(0.16,0.84),
               cdf = True,
               colors=[color_raw,color_res, color_gaus_res],
               labels=["raw","NN", "NN + Gauss BT"], xlabel="BT (K)",ylabel = "CDF (%)", 
               vlines=[(225, "225 K",0.6), (241, "241 K",0.5)], 
               invert_x=True,xlim=(191,330),
               title = "a)", legend = False)



inputlayer.plot_pdf_pair( ax=axes[1], data_list=[hist_tb["cdf_counts_tb"],hist_tb["cdf_counts_tb_res"], hist_tb["cdf_counts_tb_gaus_res"]], 
               alltimestamps = False, main = True, uncertainty = False,
               percentiles =(0.16,0.84),
               cdf = True,
               colors=[color_raw,color_res, color_gaus_res],
               labels=["raw","NN", "NN + Gauss IWP"], xlabel="BT (K)",ylabel = "CDF (%)", 
               vlines=[(225, "225 K",0.5), (241, "241 K",0.8)], 
               invert_x=True,xlim=(200,245),ylim=(0.95,1),
               title = "b)", legend = False)

inputlayer.plot_pdf_pair(ax=axes[2],
    data_list=[hist_iwp["cdf_contri_iwp"], hist_iwp["cdf_contri_iwp_res"],hist_iwp["cdf_contri_iwp_gaus_res"]],
    alltimestamps=False, main=True, uncertainty=True, percentiles=(0.16, 0.84),
    cdf=True,
    colors=[color_raw,color_res, color_iwp_obs],
    labels=["raw", "NN", "NN + Gauss"], xlabel="IWP (kg m$^{-2}$)", ylabel="Cummulative\nIWP amount (%)",
    xscale="log", xlim=(1e-2+0.001, 20),
    title="c)", legend = False)


handles, labels = [], []
seen = {}
for ax in [axes[0], axes[1]]:
    for h, l in zip(*ax.get_legend_handles_labels()):
        if l not in seen:
            seen[l] = h
            handles.append(h)
            labels.append(l)
        
fig.legend(handles, labels,
           loc="lower center",
           bbox_to_anchor=(0.5, -0.07),
           #ncol=len(labels),
            ncol=len(labels),
           frameon=False)

fig.suptitle(f"{date_label} UTC", fontweight="bold", y=0.9)
plt.tight_layout()

fig.savefig(os.path.join(output_directory,f"OBS_comparison_with_and_without_gauss_cdf_BT_{date}.pdf"), bbox_inches="tight")
plt.show()
```
