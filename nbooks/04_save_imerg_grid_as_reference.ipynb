# load IMERG precip data and save its grid
- to later load in other settings for resampling, regridding ICON to regular grid etc
- IMERG chosen as reference because resolution is lowest
- output: "../output/aux/imerg_grid.nc"

```python
import numpy as np
import xarray as xr
from pathlib import Path

# load tools
import sys 
sys.path.append("../tools")
import handle_OBS_data
```

```python
# set date
date = "2006-06-06-1800"

# load data
precip = handle_OBS_data.load_clean_crop_imerg(date,30)
```

```python
# select only grid 
grid = precip.coords.to_dataset()
grid
```

```python
# save 
output_dir =  "../output/aux/"

import os
os.makedirs(output_dir, exist_ok=True)

precip.to_netcdf(os.path.join(output_dir, "imerg_grid.nc"))
```
