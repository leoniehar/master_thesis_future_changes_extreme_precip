# load data
- IMERG and Nasa-Tb from GES DISC with EarthData
    - **Prerequisite**: Earthdata account and Credential files for earthdata (https://github.com/nasa/gesdisc-tutorials/blob/main/notebooks/How_to_Generate_Earthdata_Prerequisite_Files.ipynb)
- CCIC from AWS


```python
# load tools
import sys 
sys.path.append("../tools")
import handle_OBS_data

# set date
date = "2006-06-06-1800"

# load ddata
tb = handle_OBS_data.load_nasatb_as_ds(date)
imerg = handle_OBS_data.load_imerg_as_ds(date)
ccic = handle_OBS_data.load_ccic_as_ds(date)
```

```python
#print("**** First Load **** \n\n**Tb**:\n",tb,"\n\n\n\n","**IMERG**:\n", imerg,"\n\n\n\n","**CCIC**:\n",ccic)
```

# Unify coords names, ordering etc. 

```python
# unify coordinate names etc
### no resampling etc. done here only name changes/datetime conversion/ drop vars
### so same resampling/crop functions can be applied

imerg_clean = handle_OBS_data.allign_imerg_coords_logic_with_tb(imerg)
ccic_clean = handle_OBS_data.allign_ccic_coords_logic_with_tb(ccic)

#print("**** Cleaned **** \n\n**Tb**:\n",tb,"\n\n\n\n","**IMERG Cleaned**:\n", imerg_clean,"\n\n\n\n","**CCIC cleaned**:\n",ccic_clean)
```

# crop to tropical belt (incl. 2° buffer for resampling)

```python
# crop to tropical belt
tb_cropped = handle_OBS_data.crop_tropical_belt_buffer(tb,32)
imerg_cropped = handle_OBS_data.crop_tropical_belt_buffer(imerg_clean,32)
ccic_cropped = handle_OBS_data.crop_tropical_belt_buffer(ccic_clean,32)
```
